# K Fold Cross Validation

```python
import pandas as pd
import numpy as np
from sklearn.metrics import accuracy_score
from sklearn.linear_model import LogisticRegression
```

```python
dataset = pd.read_csv('/Users/akhileshmadyastha/Desktop/ml/Social_Network_Ads.csv')
```

```python
dataset
```

```python
X = dataset.iloc[: , :-1].values
y = dataset.iloc[: , -1].values 
```

```python
def k_fold_cross_validation( model , X , y , k = 5 ) :

    fold_size = len(X) // k 
    scores = []

    for i in range(0,k) :
        start_val = i * fold_size
        end_val = start_val + fold_size

        X_val = X[start_val : end_val]
        y_val = y[start_val : end_val]

        X_train = np.concatenate([X[:start_val] , X[end_val:]] , axis= 0 )
        y_train = np.concatenate( [y[:start_val] , y[end_val:]] , axis= 0 )

        model.fit(X_train , y_train )
        y_pred = model.predict(X_val)

        accuracy = accuracy_score(y_pred , y_val )
        print(f" Fold : {i+1} Accuracy : {accuracy} ")
        scores.append(accuracy)
    
    mean_score = np.mean(scores)
    print(f"The mean score : {mean_score}")
    

    return scores
```

```python
classifier = LogisticRegression()
```

```python
scores = k_fold_cross_validation( classifier , X , y )
print(scores)
```
